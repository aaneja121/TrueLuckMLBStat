# 15 -- Version 1.1 Prospective 2026 Snapshot Review (READ-ONLY)

> **WATERMARK: PROSPECTIVE-SNAPSHOT REVIEW NOTEBOOK -- NOT A DEVELOPMENT NOTEBOOK.**
> This notebook reviews the output of ONE already-completed Version 1.1
> prospective 2026 snapshot (`prospective/run_v1_1_2026_scoring.py`). It
> performs NO fitting, scoring, downloading, or file mutation of any kind --
> it only reads the JSON/Parquet files a completed snapshot already wrote
> under `outputs/prospective/v1_1/<snapshot>/` and
> `artifacts/prospective/v1_1/<snapshot>/`. Retrospective, non-predictive
> language only -- this is a review of what the frozen Version 1.0 system
> computed for 2026 season-to-date data, never a forecast.

```python
import json
import os
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", 60)
pd.set_option("display.width", 180)

# Overridable via environment variable for offline structure/execution testing
# against a synthetic snapshot fixture -- see
# tests/test_notebook_15_prospective_review.py. Production default reviews
# the most recently written snapshot directory name under
# outputs/prospective/v1_1 (the caller must set SNAPSHOT_NAME or point the
# env vars directly at one snapshot's own outputs/artifacts directories).
SNAPSHOT_OUTPUTS_DIR = Path(
    os.environ.get("CONTACT_LUCK_PROSPECTIVE_OUTPUTS_DIR", "../outputs/prospective/v1_1/EXAMPLE")
)
SNAPSHOT_ARTIFACTS_DIR = Path(
    os.environ.get(
        "CONTACT_LUCK_PROSPECTIVE_ARTIFACTS_DIR", "../artifacts/prospective/v1_1/EXAMPLE"
    )
)

MANIFEST_PATH = SNAPSHOT_ARTIFACTS_DIR / "manifest.json"
if MANIFEST_PATH.exists():
    print(f"SNAPSHOT FOUND at {SNAPSHOT_OUTPUTS_DIR}")
else:
    print(f"SNAPSHOT NOT FOUND at {SNAPSHOT_OUTPUTS_DIR} -- set CONTACT_LUCK_PROSPECTIVE_OUTPUTS_DIR / CONTACT_LUCK_PROSPECTIVE_ARTIFACTS_DIR.")
```

## Snapshot manifest

```python
if MANIFEST_PATH.exists():
    manifest = json.loads(MANIFEST_PATH.read_text())
    print(f"manifest_version: {manifest['manifest_version']}")
    print(f"repository_commit: {manifest['repository_commit']}")
    print(f"working_tree_clean: {manifest['working_tree_clean']}")
    print(f"prospective_season: {manifest['prospective_season']}")
    print(f"data_through_date: {manifest['data_through_date']}")
    print(f"snapshot_label: {manifest['snapshot_label']}")
    print(f"model_versions: {manifest['model_versions']}")
    print(f"score_version: {manifest['score_version']}")
    print(f"qualification_threshold_set: {manifest['qualification_threshold_set']}")
    print(f"v1_seal_verification: {manifest['v1_seal_verification']}")
else:
    print("Skipped -- no snapshot manifest found.")
```

## Public score table

```python
PUBLIC_SCORE_PATH = SNAPSHOT_OUTPUTS_DIR / "public_score.json"
if PUBLIC_SCORE_PATH.exists():
    public_score_table = pd.read_json(PUBLIC_SCORE_PATH, orient="records")
    print(f"row_count: {len(public_score_table)}")
    if "qualification_status" in public_score_table.columns:
        print(f"qualification_counts: {public_score_table['qualification_status'].value_counts().to_dict()}")
    display(public_score_table.head(10))
else:
    print("Skipped -- no public score table found.")
```

## Leaderboards

```python
FAVORABLE_PATH = SNAPSHOT_OUTPUTS_DIR / "favorable_leaderboard.json"
UNFAVORABLE_PATH = SNAPSHOT_OUTPUTS_DIR / "unfavorable_leaderboard.json"
if FAVORABLE_PATH.exists() and UNFAVORABLE_PATH.exists():
    favorable = pd.read_json(FAVORABLE_PATH, orient="records")
    unfavorable = pd.read_json(UNFAVORABLE_PATH, orient="records")
    print(f"favorable_leaderboard_rows: {len(favorable)}")
    print(f"unfavorable_leaderboard_rows: {len(unfavorable)}")
    display(favorable.head(10))
    display(unfavorable.head(10))
else:
    print("Skipped -- no leaderboards found.")
```

## Score card

```python
SCORECARD_PATH = SNAPSHOT_OUTPUTS_DIR / "scorecard.json"
if SCORECARD_PATH.exists():
    score_card = json.loads(SCORECARD_PATH.read_text())
    print(json.dumps(score_card, indent=2, default=str))
else:
    print("Skipped -- no score card found.")
```

## Coverage and schema report

```python
COVERAGE_PATH = SNAPSHOT_OUTPUTS_DIR / "coverage_and_schema_report.json"
if COVERAGE_PATH.exists():
    coverage_report = json.loads(COVERAGE_PATH.read_text())
    print(json.dumps(coverage_report, indent=2, default=str))
else:
    print("Skipped -- no coverage/schema report found.")
```

## Component status summary

```python
COMPONENT_STATUS_PATH = SNAPSHOT_OUTPUTS_DIR / "component_status_summary.json"
if COMPONENT_STATUS_PATH.exists():
    component_status = json.loads(COMPONENT_STATUS_PATH.read_text())
    print(json.dumps(component_status, indent=2, default=str))
    print("\nNOTE: any component reported here as provisional stays provisional -- this ")
    print("review notebook never re-labels a provisional component as calibrated.")
else:
    print("Skipped -- no component status summary found.")
```

## Player name resolution

```python
NAME_RESOLUTION_PATH = SNAPSHOT_OUTPUTS_DIR / "name_resolution_report.json"
if NAME_RESOLUTION_PATH.exists():
    name_report = json.loads(NAME_RESOLUTION_PATH.read_text())
    print(f"requested_batter_count: {name_report['requested_batter_count']}")
    print(f"resolved_batter_count: {name_report['resolved_batter_count']}")
    print(f"unresolved_batter_count: {name_report['unresolved_batter_count']}")
    print(f"source: {name_report['source']}")
else:
    print("Skipped -- no name resolution report found.")
```
